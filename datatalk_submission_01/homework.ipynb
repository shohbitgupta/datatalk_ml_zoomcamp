```python
mode_hp = df["horsepower"].mode()[0]          # 252.0
df["horsepower"] = df["horsepower"].fillna(mode_hp)

median_value = df["horsepower"].median().item()
print(f"Median value of 'horsepower': {median_value}")
```

```python
import numpy as np


X = df[df["origin"] == "Asia"][["vehicle_weight", "model_year"]][:7]

t_matrix = X.T
#dot product of t_matrix and X
XTX = t_matrix @ X

(a, b), (c, d) = XTX.to_numpy()
# calculate the determinant of XTX
determinant = a * d - b * c
XTX_inv = (1 / determinant) * np.array([[d, -b], [-c, a]])

print("Naive inverse of XTX:\n", XTX_inv)
```

```python
values = list([1100, 1300, 800, 900, 1000, 1100, 1200])
w = (XTX_inv @ t_matrix) @ values
w.sum().item()
```
